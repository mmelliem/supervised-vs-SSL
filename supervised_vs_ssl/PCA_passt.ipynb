```python
# for byola but confusing

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from tqdm import tqdm
import seaborn as sns
import glob

OUTPUT_DIR = '/home/mellie/supervised-vs-SSL/supervised_vs_ssl/PCA_results_byola'
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Starting PaSST Analysis...")

# Define path for PaSST features
PASST_FEATURES = '/home/mellie/supervised-vs-SSL/data/byola_features'

# Load PaSST labels
passt_labels_df = pd.read_csv('/home/mellie/supervised-vs-SSL/supervised_vs_ssl/byola_predictions.csv')
print(f"Loaded PaSST labels with {len(passt_labels_df)} entries")

# Create task labels for PaSST (0 for music, 1 for speech)
music_genres = ['Folk', 'Rock', 'unknown', 'Electronic', 'Spoken Weird', 'Disco', 'Post-Punk', 'Avant-Garde']
speech_commands = ['go', 'seven']

# Create task labels (0 for music, 1 for speech)
passt_task_labels = np.array([(1 if label in speech_commands else 0) for label in passt_labels_df['actual_label']])

# Print label distribution
print(f"Number of speech samples: {np.sum(passt_task_labels == 1)}")
print(f"Number of music samples: {np.sum(passt_task_labels == 0)}")

def process_passt_features():
    """Simplified function to process PaSST features and perform PCA"""
    # Create output directory
    passt_output_dir = os.path.join(OUTPUT_DIR, 'passt')
    os.makedirs(passt_output_dir, exist_ok=True)
    
    # Get layer directories (assuming each subdirectory is a layer)
    layer_dirs = [d for d in os.listdir(PASST_FEATURES) 
                 if os.path.isdir(os.path.join(PASST_FEATURES, d))]
    
    # Store PCA results for each layer
    layer_pca_results = {}
    layer_info = []
    
    for layer_name in layer_dirs:
        # Each subdirectory is a layer - use its path directly
        feature_dir = os.path.join(PASST_FEATURES, layer_name)
        
        print(f"Processing layer: {layer_name}")
        
        # Get all .npy files in this directory
        feature_files = glob.glob(os.path.join(feature_dir, "*.npy"))        
        
        # Load and process all features
        all_features = []
        
        for file_path in tqdm(feature_files, desc=f"Processing {layer_name}"):
            try:
                # Load features
                features = np.load(file_path)
                
                # Flatten if needed
                if features.ndim > 2:
                    features = features.reshape(features.shape[0], -1)
                elif features.ndim == 1:
                    features = features.reshape(1, -1)
                
                all_features.append(features)
            except Exception as e:
                print(f"Error processing {file_path}: {e}")
        
        if not all_features:
            print(f"No valid features processed for {layer_name}, skipping...")
            continue
        
        # Stack all features
        features = np.vstack(all_features)
        print(f"Loaded {features.shape[0]} samples with {features.shape[1]} features each")
        
        # Standardize features
        scaler = StandardScaler()
        features_scaled = scaler.fit_transform(features)
        
        # Apply PCA
        pca = PCA(n_components=2)
        features_pca = pca.fit_transform(features_scaled)
        
        # Save PCA results
        pca_dir = os.path.join(passt_output_dir, f"{layer_name}_pca")
        os.makedirs(pca_dir, exist_ok=True)
        np.save(os.path.join(pca_dir, "pca_components.npy"), features_pca)
        
        # Save explained variance
        explained_variance = pca.explained_variance_ratio_
        np.save(os.path.join(pca_dir, "explained_variance.npy"), explained_variance)
        
        # Store in dictionary for visualization
        layer_pca_results[layer_name] = features_pca
        
        # Store layer info
        layer_info.append({
            'layer': layer_name,
            'n_samples': features.shape[0],
            'n_features': features.shape[1],
            'explained_variance_pc1': explained_variance[0],
            'explained_variance_pc2': explained_variance[1],
            'total_explained_variance': sum(explained_variance)
        })
    
    # Save layer info to CSV
    pd.DataFrame(layer_info).to_csv(os.path.join(OUTPUT_DIR, "layer_info.csv"), index=False)
    
    return layer_pca_results

def plot_pca_results(layer_pca_results):
    """Plot PCA visualization for each layer"""
    if not layer_pca_results:
        print("No PCA results to plot")
        return
    
    n_layers = len(layer_pca_results)
    n_cols = min(3, n_layers)
    n_rows = (n_layers + n_cols - 1) // n_cols
    
    plt.figure(figsize=(n_cols * 6, n_rows * 5))
    
    # First, get filenames from the feature directory to align with labels
    layer_name = list(layer_pca_results.keys())[0]  # Just use first layer to get filenames
    feature_files = sorted(glob.glob(os.path.join(PASST_FEATURES, layer_name, "*.npy")))
    filenames = [os.path.basename(f).replace('.npy', '') for f in feature_files]
    
    # Map filenames to indices in the labels dataframe
    filename_to_label = {}
    for idx, row in passt_labels_df.iterrows():
        # Extract filename from the path in the CSV
        # Assuming format like '/path/to/file/filename.wav'
        csv_filename = os.path.basename(row['file_path']).replace('.wav', '')
        filename_to_label[csv_filename] = row['actual_label']
    
    # Create label array aligned with our feature files
    aligned_labels = []
    for filename in filenames:
        if filename in filename_to_label:
            label = filename_to_label[filename]
            # Apply task classification (0=music, 1=speech)
            is_speech = 1 if label in speech_commands else 0
            aligned_labels.append(is_speech)
        else:
            # If filename not found in labels, default to unknown (mark as -1)
            aligned_labels.append(-1)
    
    aligned_labels = np.array(aligned_labels)
    print(f"Created aligned labels array with {len(aligned_labels)} entries")
    print(f"Speech samples: {np.sum(aligned_labels == 1)}")
    print(f"Music samples: {np.sum(aligned_labels == 0)}")
    print(f"Unknown samples: {np.sum(aligned_labels == -1)}")
    
    for i, (layer_name, features_pca) in enumerate(layer_pca_results.items()):
        # Create subplot
        plt.subplot(n_rows, n_cols, i+1)
        
        # Create a colormap
        cmap = sns.color_palette("coolwarm", as_cmap=True)
        
        # Plot points colored by aligned task labels
        sc = plt.scatter(features_pca[:, 0], 
                   features_pca[:, 1],
                   c=aligned_labels, 
                   cmap=cmap,
                   alpha=0.7, 
                   s=30)
        
        plt.title(f'Layer: {layer_name}')
        plt.colorbar(sc, label='Task (0=Music, 1=Speech)')
        plt.xlabel('PC1')
        plt.ylabel('PC2')
    
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, 'byola_pca_visualization.png'), dpi=300)
    plt.show()

# Run the analysis
print("Processing features and applying PCA...")
layer_pca_results = process_passt_features()

# Visualize the results
print("Creating PCA visualizations...")
plot_pca_results(layer_pca_results)
```
